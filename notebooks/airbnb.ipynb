### Type -  int64 & float64

```python
# Steps -
# - Check missing percentage
# - If 50 % or above data is missing then drop the column if it is not important
# - Or,
# - Check outlier in each column
# - If outlier present , replace NaN values with median
# - If outlier is not present, replace NaN values with mean
```

```python
# Check missing percentage
for column in df.columns:
    if df[column].dtypes == "int64" or df[column].dtypes =="float64":
        missing_pct = np.trunc(df[column].isnull().sum()/len(df))*100
        print(column, "(", df[column].isnull().sum(), ",", missing_pct,"%", ")")
```

```python
# Function of checking outliers
def is_outlier(col: str, df: pd.DataFrame) -> bool:

    Q1 = df[col].quantile(0.25) # Lower quantile -> 25 %
    Q2 = df[col].median()       # Median -> 50 %
    Q3 = df[col].quantile(0.75) # Upper auantile -> 75 %

    IQR = Q3 - Q1 # IQR

    lower_limit = Q1 - (IQR * 1.5)
    upper_limit = Q3 + (IQR * 1.5)

    if len(df.loc[df[col] < lower_limit]) > 0 or len(df.loc[df[col] > upper_limit]) > 0:
        return True
    else:
        return False
```

```python
# Fixing nulls
for column in df.columns:
    if df[column].dtypes == "int64" or df[column].dtypes =="float64":
        missing_pct = np.trunc(df[column].isnull().sum()/len(df))*100

        if missing_pct <= 50:
            if is_outlier(col=column, df=df):
                df[column] = df[column].replace(np.nan, df[column].median())
            else:
                df[column] = df[column].replace(np.nan, df[column].mean())
```

```python
# Final check
for column in df.columns:
    if df[column].dtypes == "int64" or df[column].dtypes =="float64":
        missing_pct = np.trunc(df[column].isnull().sum()/len(df))*100
        print(column, "(", df[column].isnull().sum(), ",", missing_pct,"%", ")")
```

### Type - object

```python
# Steps -
# - Check missing percentage
# - If 50 % or above data is missing then drop the column if it is not important
# - Or,
# - Fill it with 'unknown'
# - If missing percentage <= 10, fill NaN with most frequent value of the column using mode
```

```python
# Check missing percentage
for column in df.columns:
    if df[column].dtypes == "object":
        missing_pct = np.trunc(df[column].isnull().sum()/len(df))*100
        print(column, "(", df[column].isnull().sum(), ",", missing_pct,"%", ")")
```

```python
for column in df.columns:
    if df[column].dtypes == "object":
        missing_pct = np.trunc(df[column].isnull().sum()/len(df))*100

        if missing_pct <= 10:
            df[column] = df[column].replace(np.nan, df[column].mode()[0])
        else:
            df[column].fillna('unknown', inplace=True)
```

```python
# - Find outliers from int or float type columns
# - If outliers present then replace them with lower or upper limit values
# - Or keep them as it is if they are logical by having outliers
```

```python
# Check for outliers
for column in df.columns:
    if df[column].dtypes == "int64" or df[column].dtypes == "float64":
        if is_outlier(col=column, df=df):
            print(column)
```

```python
# Check if they should contain outliers or not
df[['lat', 'long', 'minimum_nights', 'number_of_reviews']]
```

```python
# Column = 'lat'

Q1 = df['lat'].quantile(0.25) # Lower quantile -> 25 %
Q2 = df['lat'].median()       # Median -> 50 %
Q3 = df['lat'].quantile(0.75) # Upper auantile -> 75 %

IQR = Q3 - Q1 # IQR

lower_limit = Q1 - (IQR * 1.5)
upper_limit = Q3 + (IQR * 1.5)

lat_upper_mask = df['lat'] > upper_limit
print(df.loc[lat_upper_mask, 'lat'])

lat_lower_mask = df['lat'] < lower_limit
print(df.loc[lat_lower_mask, 'lat'])

lat_inner_mask = ((df['lat'] >= lower_limit) & (df['lat'] <= upper_limit))
print(df.loc[lat_inner_mask, 'lat'])
```

```python
# column = 'long'

Q1 = df['long'].quantile(0.25) # Lower quantile -> 25 %
Q2 = df['long'].median()       # Median -> 50 %
Q3 = df['long'].quantile(0.75) # Upper auantile -> 75 %

IQR = Q3 - Q1 # IQR

lower_limit = Q1 - (IQR * 1.5)
print(lower_limit)

upper_limit = Q3 + (IQR * 1.5)
print(upper_limit)

long_upper_mask = df['long'] > upper_limit
print(df.loc[long_upper_mask, 'long'])

long_lower_mask = df['long'] < lower_limit
print(df.loc[long_lower_mask, 'long'])

long_inner_mask = ((df['long'] >= lower_limit) & (df['long'] <= upper_limit))


print(df.loc[long_inner_mask, 'long'])
```

```python
# Column = 'minimum_nights'

Q1 = df['minimum_nights'].quantile(0.25) # Lower quantile -> 25 %
Q2 = df['minimum_nights'].median()       # Median -> 50 %
Q3 = df['minimum_nights'].quantile(0.75) # Upper auantile -> 75 %

IQR = Q3 - Q1 # IQR

# Lower limit
lower_limit = Q1 - (IQR * 1.5)
print(lower_limit)

# upper limit
upper_limit = Q3 + (IQR * 1.5)
print(upper_limit)

# values which is greater than upper limit in column minimum_nights
print(f"values greater than : {upper_limit}")
nights_upper_mask = df['minimum_nights'] > upper_limit
print(df.loc[nights_upper_mask, 'minimum_nights'])

# values which is lesser than lower limit in column minimum_nights
print(f"Values lesser than {lower_limit}")
nights_lower_mask = df['minimum_nights'] < lower_limit
print(df.loc[nights_lower_mask, 'minimum_nights'])

# Values which falls under the lower and upper range
nights_inner_mask = ((df['minimum_nights'] >= lower_limit) & (df['minimum_nights'] <= upper_limit))
print(df.loc[nights_inner_mask, 'minimum_nights'])

# Replacing negative values with median
df.loc[nights_lower_mask, 'minimum_nights'] = df['minimum_nights'].median()
print(df.loc[nights_lower_mask, 'minimum_nights'])
```
