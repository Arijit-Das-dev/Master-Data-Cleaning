```python
# Check missing percentage
for column in df.columns:
    if df[column].dtypes == "int64" or df[column].dtypes =="float64":
        missing_pct = np.trunc(df[column].isnull().sum()/len(df))*100
        print(column, "(", df[column].isnull().sum(), ",", missing_pct, ")")
```
