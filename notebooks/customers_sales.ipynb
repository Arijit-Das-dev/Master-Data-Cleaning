```python
# 'gender'
print(df['gender'].value_counts())
print(df['gender'].unique())

# Fix
df['gender'] = df['gender'].replace(
    {
        "m ":"male",
        "M":"male",
        "MALE":"male",
        "F":"female",
        "FEMALE":"female",
        "f ":"female"
    }
)

# Final check
print(df['gender'].value_counts())
print(df['gender'].unique())
```

```python
# city
print(df['city'].value_counts())
print(df['city'].unique())

# Fix
df['city'] = (
    df['city']
    .str.lower()
    .str.strip()
)

# Check
print(df['city'].value_counts())
print(df['city'].unique())
```

```python
# signup_date
print(df['signup_date'].dtypes)

df['signup_date'] = pd.to_datetime(df['signup_date'], errors='coerce')
print(df['signup_date'].dtypes)
```

```python
# last_purchase_date
print(df['last_purchase_date'].dtypes)
df['last_purchase_date'] = pd.to_datetime(df['last_purchase_date'], errors='coerce')
print(df['last_purchase_date'].dtypes)
```

```python
# email
# email type : abc@gmail.com / abc12@gmail.com
# Basic cleaning

df['email'] = (
    df['email']
    .str.lower()
    .str.strip()
    .str.replace(r'[^0-9a-z@.]', '', regex=True)
)
```

```python
# country
print(df['country'].value_counts())
print(df['country'].unique())

# Fix
valid_pattern = {
    "India":"india",
    "IND":"india",
    "InDia":"india"
}

df['country'] = df['country'].map(valid_pattern)

# final check
print(df['country'].value_counts())
print(df['country'].unique())
```

```python
# 'age'
df['age'] = df['age'].str.lower()                              # to lower case
df['age'] = df['age'].str.replace(r'[a-z\s]', '', regex=True)  # replacing letters and spaces with empty string
df['age'] = df['age'].replace('', np.nan)                      # replacing empty string with NaN (Nulls)
df['age'] = df['age'].replace(np.nan, df['age'].mode()[0])     # replacing nulls with most frequent values
df['age'] = df['age'].astype(float).astype(int)                # converting data type from str to float to int
```

```python
# Now int or float type columns
for column in df.columns:
    if df[column].dtypes == "int64" or df[column].dtypes == "float64":
        print(column)
```

```python
# Check outliers
def is_outlier(col: str, df: pd.DataFrame) -> bool:

    Q1 = df[col].quantile(0.25)
    Q2 = df[col].median()
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - (IQR * 1.5)
    upper_limit = Q3 + (IQR * 1.5)

    if len(df.loc[df[col] > upper_limit]) > 0 or len(df.loc[df[col] < lower_limit]) > 0:
        return True
    else:
        return False
```

```python
# purchase_amount
print((df[['purchase_amount', 'feedback_score']].isnull().sum()/len(df[['purchase_amount', 'feedback_score']]))*100)

# Fix
columns = ['purchase_amount', 'feedback_score']

for cols in columns:
    if is_outlier(col=cols, df=df):
        df[cols] = df[cols].replace(np.nan, df[cols].median())
    else:
        df[cols] = df[cols].replace(np.nan, df[cols].mean())

# Final check
print((df[['purchase_amount', 'feedback_score']].isnull().sum()/len(df[['purchase_amount', 'feedback_score']]))*100)
```

```python
# phone number
# For data cleaning
# If you're cleaning the dataset, don't treat phone number as an integer.

# To check in phone numbers
# This is important because phone numbers can have:
"""
# leading zeros
# country codes (+91)
# spaces
# hyphens
# parentheses
"""

# convert it's data type for better analysis
df['phone_number'] = df['phone_number'].astype(str)

# Leading zeros
zeros_mask = df['phone_number'].str.startswith('0', na=False)
print(zeros_mask.sum())

# country code (+91)
country_code_mask = df['phone_number'].str.startswith('+91', na=False)
print(country_code_mask.sum())

# length > 10
greater_than_ten_mask = df['phone_number'].str.len() > 10
print(greater_than_ten_mask.sum())


df['phone_number'] = df['phone_number'].astype(int)
```

# 6 - fix nulls in all type of columns

```python
# Object type columns
for column in df.columns:
    if df[column].dtypes == "object":
        if column == "email":
            continue
        else:
            missing_pct = (df[column].isnull().sum()/len(df[column]))*100
            if missing_pct <= 10:
                df[column] = df[column].replace(np.nan, df[column].mode()[0])
            else:
                df[column] = df[column].replace(np.nan, 'unknown')
```

```python
# Data type columns
for column in df.columns:
    if df[column].dtypes == '<M8[ns]':
        missing_pct = (df[column].isnull().sum()/len(df[column]))*100

        if missing_pct <= 50:
            df[column] = df[column].ffill()
```

```python
# final check
df.isnull().sum()
```
