```python
# IMPORTING ESSENTIAL LIBRARIES
import numpy as np
import pandas as pd
```

```python
# LOAD THE DATASET
df = pd.read_csv(r"C:\Users\Arijit\Downloads\marketing_campaign_data_messy.csv")
df
```

```python
# EXPLORING THE DATASET
print("Total rows : ", df.shape[0])
print("Total columns : ", df.shape[1])
print("Data dimension : ", df.ndim)
print("Data size : ", df.size)

print()

# SUMMARY STATISTICS
df.info()
```

```python
# CLEANING COLUMNS
# IDENTIFYING COLUMNS WHICH CONTAINS SPACES, SPECIAL CHARACTERS, LOWER-CASE, UPPER-CASE.

print(df.columns[df.columns == r'[0-9]'])               # CHECKING FOR DIGITS FROM 0 TO 9
print(df.columns[df.columns != df.columns.str.strip()]) # CHECKING IF ANY COLUMN CONTAINS SPACES
print(df.columns[df.columns != df.columns.str.lower()]) # CHECKING IF ANY COLUMN CONTAINS UPPER CASE

# CONVERTING EACH COLUMNS TO IT'S LOWER CASE
df.columns = df.columns.str.lower() # CONVERTING EACH COLUMN NAMES TO LOWER CASE
df.columns = df.columns.str.strip() # REMOVING SPACES FROM EACH COLUMN

df.columns
```

```python
# Check list :

# Contains any non-digit	            df['spend'].str.contains(r'[^0-9]', na=False)
# Contains only digits	                df['spend'].str.fullmatch(r'\d+', na=False)
# Contains letters	                    df['spend'].str.contains(r'[A-Za-z]', na=False)
# Contains special characters	        df['spend'].str.contains(r'[^A-Za-z0-9]', na=False)


# CHECKING COLUMNS
contains_dollar = df['spend'].astype(str).str.contains(r'\$', na=False)
contains_spaces = df['spend'].astype(str).str.contains(r' ', na=False)
contains_only_digits = df['spend'].astype(str).str.fullmatch(r'\d+', na=False)
contains_letters = df['spend'].astype(str).str.contains(r'[A-Za-z]', na=False)
contains_special_characters = df['spend'].astype(str).str.contains(r'[^A-Za-z0-9]', na=False)

print("Contains $ :", contains_dollar.sum())
print("Contains spaces :", contains_spaces.sum())
print("Contains only digits :", contains_only_digits.sum())
print("Contains letters :", contains_letters.sum())
print("Contains special characters :", contains_special_characters.sum())

print()

# REMOVING $
df['spend'] = df['spend'].str.replace('$', "")
contains_dollar = df['spend'].astype(str).str.contains(r'\$', na=False)
print("Contains $ : ", contains_dollar.sum())
```

```python
# IMPROVING CATAGORICAL TYPO

# CHEACKING HOW MANY UNIQUE VALUES ARE PRESENT
print(df['channel'].value_counts()) # OCCURRENCE OF EACH UNIQUE VALUE
print()
print(df['channel'].unique()) # EACH UNIQUE VALUES
print()
print("\nTotal unique values : ", df['channel'].nunique()) # NUMBER OF UNIQUE VALUES

# CLEANING TYPOS
df['channel'] = df['channel'].replace({
    "Tik_Tok":"TikTok",
    "Facebok":"Facebook",
    "E-mail":"Email",
    "Gogle":"Google Ads",
    "Insta_gram":"Instagram"  
})

print(df['channel'].value_counts())
print()
print(df['channel'].unique())
print()
print("Total unique values : ", df['channel'].nunique())
```

```python
# FIXING BOOLEAN COLUMNS

print(df['active'].value_counts()) # NUMBER OF OCCURRENCE OF EACH VALUES 
print(df['active'].unique())       # EACH UNIQUE VALUES
print(df['active'].nunique())      # NUMBER OF UNIQUE VALUES

df['active'] = df['active'].replace(
    {
        "1":"True",
        "Yes":"True",
        "Y":"True",
        "0":"False",
        "No":"False"
})

print(df['active'].value_counts())
print(df['active'].unique())
print(df['active'].nunique())
```

### DATE PARSING

```python
# DATE PARSING

print(df['start_date'].dtype)

df['start_date'] = pd.to_datetime(df['start_date'], errors='coerce')
df['end_date'] = pd.to_datetime(df['end_date'], errors='coerce')

df['start_date'].dtypes
df['end_date'].dtypes
```

```python
# DROPPING DUPLICATE COLUMNS
# THERE ARE TWO DUPLICATED COLUMNS => clicks
df = df.loc[:, ~df.columns.duplicated()]
df
```

### LOGICAL INTIGRITY

```python
# NOW WE HAVE TO COMPARE CLICKS WITH IMPRESSIONS
# IMPRESSION = How many times a object shown or displayed on the screen
# CLICKS = How many times someone clicking that object
# IT IS FIXED THAT IMPRESSIONS >= CLICKS.
# EXAMPLE : IMPRESSION = 100 , CLICK = 50. WE CAN UNDERSTAND THAT 50 TIMES THAT OBJECT IS CLICKED OUT OF 100

impossible_mask = df['clicks'] > df['impressions']
df.loc[impossible_mask, ['clicks', 'impressions']]
```

```python
# IT IS SIMPLE LOGIC THAT START DATE WILL ALWAYS BE LESS THAN END DATE.
# GETTING THOSE ROWS WHERE END DATE IS LESS THAN START DATE

df.loc[df['end_date'] < df['start_date']]
```

```python
# HOW MANY ROWS ARE AFFECTED
time_travel_mask = df['end_date'] < df['start_date']
df.loc[time_travel_mask, ['end_date', 'start_date']]

# FIXING 
df.loc[time_travel_mask, 'end_date'] = df.loc[time_travel_mask, 'start_date'] + pd.Timedelta(days=30)
```

```python
# FIXED
df.loc[df['end_date'] < df['start_date']]
```

```python
# OUTLIER TREATMENT

# CONVERTING DATA TYPES
df['spend'] = df['spend'].astype(str).astype(float).astype(int)
df['spend'].dtypes

Q1 = df['spend'].quantile(0.25)
Q3 = df['spend'].quantile(0.75)

IQR = Q3-Q1

upper_limit = Q3 + (IQR * 1.5)
lower_limit = Q1 - (IQR * 1.5)

print(upper_limit)

df.loc[df['spend'] > upper_limit] = upper_limit
df.loc[df['spend'] < lower_limit] = lower_limit
```

```python
# CHECKING IF OUTLIER STILL PRESENT OR NOT.
# FIXED
print(df.loc[df['spend'] > upper_limit])
print(df.loc[df['spend'] < lower_limit])
```
